```python
import pandas as pd

# Cargamos el contenido del archivo en la variable df
df = pd.read_csv('cafe_sales.csv')

df.head(5)
```

Hallazgos del archivo inicial

```python
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
```

Sustitución de valores "ERROR" y "UNKNOWN" por NaN

```python
df = df.replace(["ERROR", "UNKNOWN"], pd.NA)
print(df.info())
print(df.isna().sum())
```

Limpieza de Transaction ID

```python
df["Transaction ID"].isna().sum()
```

Limpieza de Item

```python
df["Item"].isna().sum()
df = df.dropna(subset=["Item"])

df["Item"].isna().sum()
```

```python
print(df.isna().sum())
```

Limpieza Quantity

```python
# convertir la columna "Quantity" a tipo numérico
df["Quantity"] = pd.to_numeric(df["Quantity"], errors='coerce')

# eliminar nulos
df = df.dropna(subset=["Quantity"])

# convertir Quantity a entero
df["Quantity"] = df["Quantity"].astype(int)

print(df.isna().sum())
df.info()
```

Limpieza de Price Per Unit

```python
# convertir la columna "Quantity" a tipo numérico
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors='coerce')

# eliminar nulos
df = df.dropna(subset=["Price Per Unit"])

print(df.isna().sum())
df.info()
```

Limpieza de Total Spent

```python
df["Total Spent"] = pd.to_numeric(df["Total Spent"], errors='coerce')
df.head(25)

# encontrar el valor de Total Spent
df["Total Spent"] = df["Total Spent"].fillna(
    df["Quantity"] * df["Price Per Unit"]
)
```

```python
print(df.isna().sum())
df.info()
```

Limpieza de Payment Method

```python
# eliminar nulos

df["Payment Method"].value_counts(dropna=False)
df = df.dropna(subset=["Payment Method"])
```

Limpieza de Location

```python
df["Location"].value_counts(dropna=False)

# eliminar nulos
df = df.dropna(subset=["Location"])
df["Location"].value_counts(dropna=False)
```

```python
print(df.isna().sum())
df.info()
```

Limpieza de Transaction Date

```python
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"], 
    errors='coerce'
)

# eliminar nulos
df = df.dropna(subset=["Transaction Date"])
```

```python
df.info()
```

Eliminar duplicados

```python
df = df.drop_duplicates()
df.info()
```

Validar limpieza del DataFrame

```python
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
df.describe()
```

Carga de datos limpios

```python
df.to_csv('cafe_sales_clean.csv', index=False)
```
